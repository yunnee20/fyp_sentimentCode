```python
from PIL import Image, ImageDraw, ImageFont

def generate_receipt(data):
    width = 215
    height = 474

    container = 181
    padding = 20

    img = Image.new("L", (width, height), 255)
    draw = ImageDraw.Draw(img)

    # Fonts
    font_title = ImageFont.truetype("mat/jbm_extrabold.ttf", 16)
    font_sub = ImageFont.truetype("mat/jbm_bold.ttf", 12)
    font_body = ImageFont.truetype("mat/jbm_regular.ttf", 10)

    y = 20

    # ----------------------
    # TITLE BLOCK
    # ----------------------
    # black background block
    draw.rectangle((padding, y, width-padding, y+96), fill=0)

    # white text
    draw.text((35, y+15), "WHEN AI CLAIMS", font=font_title, fill=255)
    draw.text((35, y+35), "TO KNOW US TOO", font=font_title, fill=255)
    draw.text((70, y+55), "CLEARLY", font=font_title, fill=255)

    y += 105

    # ----------------------
    # SCORE
    # ----------------------
    draw.line((padding, y, container+10, y), fill=0, width=2)
    y+= 10
    draw.text((20, y), f"SCORE: {data['score']}%  SUBOPTIMAL", font=font_sub, fill=0)
    y += 30
    draw.line((padding, y, container+10, y), fill=0, width=2)

    y += 10

    # ----------------------
    # DATA BLOCK
    # ----------------------
    lineheight = 18
    draw.text((20, y), f"Valence: {data['valence']}", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Thinking: {data['thinking']}", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Arousal: {data['arousal']}", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Anxious: {data['anxious']}", font=font_body, fill=0)

    y += 25
    draw.line((padding, y, container+10, y), fill=0)

    y += 10

    draw.text((20, y), f"Face Score: {data['face']}%", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Voice Score: {data['voice']}%", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Text Score: {data['text']}%", font=font_body, fill=0)

    y += 25
    draw.line((padding, y, container+10, y), fill=0, width= 3)
    y += 10
    draw.text((20, y), "Practice Required", font=font_sub, fill=0)
    y += 25
    draw.line((padding, y, container+10, y), fill=0)
    y += 10
    draw.text((30, y), "thank you for visiting...", font=font_body, fill=100)

    # Barcode
    y+= 20
    barcode_img = Image.open("mat/barcode.png")
    barcode_img = barcode_img.resize((150, 50))
    img.paste(barcode_img, (30, y))
    y += 55
    
    #ID code
    draw.text((35, y), "2026 05 23 2016789 1201", font=font_body, fill=0)
    #change to timestamp
    
    return img

data = {
    "score": 47,
    "valence": 0,
    "thinking": 17.6,
    "arousal": 56.3,
    "anxious": 70.2,
    "face": 35.7,
    "voice": 60,
    "text": 80
}

img = generate_receipt(data)
img.show()
```
