```python
from PIL import Image, ImageDraw, ImageFont

def generate_receipt(data):
    width = 215
    height = 474

    container = 181
    padding = 20

    img = Image.new("L", (width, height), 255)
    draw = ImageDraw.Draw(img)

    # Fonts
    font_title = ImageFont.truetype("mat/jbm_extrabold.ttf", 16)
    font_sub = ImageFont.truetype("mat/jbm_bold.ttf", 12)
    font_body = ImageFont.truetype("mat/jbm_regular.ttf", 10)

    y = 20

    # ----------------------
    # TITLE BLOCK
    # ----------------------
    # black background block
    draw.rectangle((padding, y, width-padding, y+96), fill=0)

    # white text
    draw.text((35, y+15), "WHEN AI CLAIMS", font=font_title, fill=255)
    draw.text((35, y+35), "TO KNOW US TOO", font=font_title, fill=255)
    draw.text((70, y+55), "CLEARLY", font=font_title, fill=255)

    y += 105

    # ----------------------
    # SCORE
    # ----------------------
    draw.line((padding, y, container+10, y), fill=0, width=2)
    y+= 10
    draw.text((20, y), f"SCORE: {data['score']}%  SUBOPTIMAL", font=font_sub, fill=0)
    y += 30
    draw.line((padding, y, container+10, y), fill=0, width=2)

    y += 10

    # ----------------------
    # DATA BLOCK
    # ----------------------
    lineheight = 18
    draw.text((20, y), f"Valence: {data['valence']}", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Thinking: {data['thinking']}", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Arousal: {data['arousal']}", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Anxious: {data['anxious']}", font=font_body, fill=0)

    y += 25
    draw.line((padding, y, container+10, y), fill=0)

    y += 10

    draw.text((20, y), f"Face Score: {data['face']}%", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Voice Score: {data['voice']}%", font=font_body, fill=0)
    y += lineheight
    draw.text((20, y), f"Text Score: {data['text']}%", font=font_body, fill=0)

    y += 25
    draw.line((padding, y, container+10, y), fill=0, width= 3)
    y += 10
    draw.text((20, y), "Practice Required", font=font_sub, fill=0)
    y += 25
    draw.line((padding, y, container+10, y), fill=0)
    y += 10
    draw.text((30, y), "thank you for visiting...", font=font_body, fill=100)

    # Barcode
    y+= 20
    barcode_img = Image.open("mat/barcode.png")
    barcode_img = barcode_img.resize((150, 50))
    img.paste(barcode_img, (30, y))
    y += 55
    
    #ID code
    draw.text((35, y), "2026 05 23 2016789 1201", font=font_body, fill=0)
    #change to timestamp
    
    return img

data = {
    "score": 47,
    "valence": 0,
    "thinking": 17.6,
    "arousal": 56.3,
    "anxious": 70.2,
    "face": 35.7,
    "voice": 60,
    "text": 80
}

img = generate_receipt(data)
img.show()
```

```python
from PIL import Image, ImageDraw, ImageFont
from datetime import datetime

DPI = 300
WIDTH_IN = 1.97
HEIGHT_IN = 3.94

width = int(WIDTH_IN * DPI)      # 591
height = int(HEIGHT_IN * DPI)    # 1182

scale = width / 215  # scale from your original design

def S(v):
    return int(v * scale)

def generate_receipt(data):
    img = Image.new("L", (width, height), 255)
    draw = ImageDraw.Draw(img)

    padding = S(20)
    container = S(181)

    font_title = ImageFont.truetype("mat/jbm_extrabold.ttf", S(16))
    font_sub = ImageFont.truetype("mat/jbm_bold.ttf", S(12))
    font_body = ImageFont.truetype("mat/jbm_regular.ttf", S(10))

    y = S(20)

    draw.rectangle((padding, y, width-padding, y+S(96)), fill=0)

    draw.text((S(35), y+S(15)), "WHEN AI CLAIMS", font=font_title, fill=255)
    draw.text((S(35), y+S(35)), "TO KNOW US TOO", font=font_title, fill=255)
    draw.text((S(70), y+S(55)), "CLEARLY", font=font_title, fill=255)

    y += S(105)

    draw.line((padding, y, container+S(10), y), fill=0, width=S(2))
    y += S(10)

    draw.text((S(20), y), f"SCORE: {data['score']}%  SUBOPTIMAL", font=font_sub, fill=0)

    y += S(30)
    draw.line((padding, y, container+S(10), y), fill=0, width=S(2))
    y += S(10)

    lineheight = S(18)

    draw.text((S(20), y), f"Valence: {data['valence']}", font=font_body, fill=0)
    y += lineheight
    draw.text((S(20), y), f"Thinking: {data['thinking']}", font=font_body, fill=0)
    y += lineheight
    draw.text((S(20), y), f"Arousal: {data['arousal']}", font=font_body, fill=0)
    y += lineheight
    draw.text((S(20), y), f"Anxious: {data['anxious']}", font=font_body, fill=0)

    y += S(25)
    draw.line((padding, y, container+S(10), y), fill=0)

    y += S(10)

    draw.text((S(20), y), f"Face Score: {data['face']}%", font=font_body, fill=0)
    y += lineheight
    draw.text((S(20), y), f"Voice Score: {data['voice']}%", font=font_body, fill=0)
    y += lineheight
    draw.text((S(20), y), f"Text Score: {data['text']}%", font=font_body, fill=0)

    y += S(25)
    draw.line((padding, y, container+S(10), y), fill=0, width=S(3))

    y += S(10)
    draw.text((S(20), y), "Practice Required", font=font_sub, fill=0)

    y += S(25)
    draw.line((padding, y, container+S(10), y), fill=0)

    y += S(10)
    draw.text((S(30), y), "thank you for visiting...", font=font_body, fill=100)

    y += S(20)

    barcode_img = Image.open("mat/barcode.png").convert("L")
    barcode_img = barcode_img.resize((S(150), S(50)))
    img.paste(barcode_img, (S(30), y))

    y += S(55)

    timestamp = datetime.now().strftime("%Y %m %d %H%M%S")
    draw.text((S(35), y), timestamp, font=font_body, fill=0)

    return img


data = {
    "score": 47,
    "valence": 0,
    "thinking": 17.6,
    "arousal": 56.3,
    "anxious": 70.2,
    "face": 35.7,
    "voice": 60,
    "text": 80
}

img = generate_receipt(data)

# Save high-res PNG
img.save("receipt.png", dpi=(DPI, DPI))

# Save as PDF at correct print size
img_rgb = img.convert("RGB")
img_rgb.save("receipt.pdf", resolution=DPI)
```

```python
import os
import platform
import subprocess
from datetime import datetime

def save_and_print_receipt(data):
    img = generate_receipt(data)

    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    filename = f"receipt_{timestamp}.pdf"

    img_rgb = img.convert("RGB")
    img_rgb.save(filename, resolution=300)

    system = platform.system()

    if system == "Windows":
        os.startfile(filename, "print")

    else:
        print("Unsupported operating system")

    print(f"Printed: {filename}")
```

```python
data = {
    "score": 47,
    "valence": 0,
    "thinking": 17.6,
    "arousal": 56.3,
    "anxious": 70.2,
    "face": 35.7,
    "voice": 60,
    "text": 80
}

save_and_print_receipt(data)
```
